# skl Sample Datasets — Export

This notebook exports the sample datasets shipped in `datasets/` using
scikit-learn's bundled loaders. It replaces the former `main.py` driver: the
notebook is both the executable source of truth and the documentation for how
each CSV is produced.

Two datasets are exported:

- `datasets/regression_diabetes.csv` — the diabetes regression dataset.
- `datasets/classification_iris.csv` — the iris classification dataset.


```python
from pathlib import Path

import pandas as pd
from sklearn.datasets import load_diabetes, load_iris

OUTPUT_DIR = Path("datasets")


def export_regression_dataset() -> Path:
    """Export the diabetes regression dataset to CSV."""
    dataset = load_diabetes(as_frame=True)

    df = dataset.data.copy()
    df["target"] = dataset.target

    output_path = OUTPUT_DIR / "regression_diabetes.csv"
    df.to_csv(output_path, index=False)

    return output_path


def export_classification_dataset() -> Path:
    """Export the iris classification dataset to CSV."""
    dataset = load_iris(as_frame=True)

    df = dataset.data.copy()
    df["target"] = dataset.target

    output_path = OUTPUT_DIR / "classification_iris.csv"
    df.to_csv(output_path, index=False)

    return output_path
```

## 1. Export the datasets

Both exports write into `datasets/`, creating the directory if it does not yet
exist. Re-running this cell overwrites the committed CSVs with the same
content, so it doubles as a regeneration step.

```python
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

regression_path = export_regression_dataset()
classification_path = export_classification_dataset()

print(f"Regression dataset:     {regression_path}")
print(f"Classification dataset: {classification_path}")
```

## 2. Verify the exports

Read the freshly written CSVs back and report their shape and columns, so a
failed or partial export is obvious from the notebook output alone.

```python
for path in (regression_path, classification_path):
    df = pd.read_csv(path)
    print(f"\n{path}")
    print(f"Rows: {len(df)}, features: {len(df.columns) - 1}")
    print(f"Columns: {df.columns.tolist()}")
```

Finally, confirm the classification labels are the three iris species encoded
as `0`, `1`, `2`.

```python
iris = pd.read_csv(classification_path)
print(f"Classification labels: {sorted(iris['target'].unique())}")
```
